```python
import tensorflow as tf
import numpy as np

print("=== 1. CREATION DYAL TENSORS (Asasiyat) ===")

# 1. Tensor 3adi (Constant - maytbeddelch)
t_const = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
print("Tensor Constant:\n", t_const.numpy())

# 2. Mn NumPy l'Tensor (Bzaf katkhdem biha)
data_np = np.array([10, 20, 30])
t_from_np = tf.convert_to_tensor(data_np, dtype=tf.float32)
print("\nTensor mn NumPy:", t_from_np.numpy())


print("\n=== 2. MANIPULATION (Tbdel l'forma d Tensor) ===")

# Reshape: nbdlo chkel mn 1D l 2D (Matrice)
t_1d = tf.constant([1, 2, 3, 4, 5, 6])
t_2d = tf.reshape(t_1d, [2, 3]) # 2 stoura, 3 a3mida
print("Reshaped (2x3):\n", t_2d.numpy())


print("\n=== 3. TRAITEMENT DYAL TSWIRA (Hada howa sa7 f Live Coding) ===")

# Kantsaylo blli 3ndna tswira fiha 28x28 pixels w 3 couleurs (RGB)
# F l7a9i9a ghatjibha b tf.keras.preprocessing.image.load_img wla tf.io
image_brute = tf.random.uniform(shape=[28, 28, 3], minval=0, maxval=255, dtype=tf.int32)
image_brute = tf.cast(image_brute, tf.uint8) # Pixels dima kaykouno uint8 f lwel

print("Shape 9bel:", image_brute.shape) # Ghat3tik (28, 28, 3)
print("Type 9bel:", image_brute.dtype)   # Ghat3tik uint8

# ETAPE A: Normalisation (Nredoha bin 0 w 1, w type twlli float32)
image_norm = tf.cast(image_brute, tf.float32) / 255.0

print("\nType be3d Normalisation:", image_norm.dtype) # Ghat3tik float32
print("A3la 9ima dyal pixel daba:", tf.reduce_max(image_norm).numpy()) # Ghatkoun sgher ola katsawi 1

# ETAPE B: Nzado l'Batch Dimension (Daroooooori 9bel mat3tiha l model)
image_batch = tf.expand_dims(image_norm, axis=0)

print("\nShape be3d expand_dims (Batch):", image_batch.shape) 
# Daba shape hiya (1, 28, 28, 3) -> Ya3ni Lot fih 1 tswira, tol 28, 3erd 28, 3 couleurs


print("\n=== 4. TEST M3A MODEL (Kifash Model kayakhod data) ===")

# Nsaybo model sghir dyal Deep Learning bach njerbo (Classification)
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28, 3)), # Kansr7o tswira
    tf.keras.layers.Dense(10, activation='relu'),     # Couche mkhabya
    tf.keras.layers.Dense(2, activation='softmax')    # 2 classes (Matalan: Chat ola Chien)
])

# N3tiweh tswira li wjjdna (Dik li derna liha expand_dims)
prediction = model(image_batch)

print("Prediction shape:", prediction.shape) # Ghatakhrej (1, 2) -> 1 tswira, 2 probabilites
print("Resultat dyal prediction (Probabilites):", prediction.numpy())
```
